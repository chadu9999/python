```python
for i in range(1, 5+1):
    for j in range(i):  # print stars on left
        print("*", end=" ")
    for k in range(2 * (5 - i)):  # print space in middle
        print(" ", end=" ")
    for j in range(i):  # print stars on right
        print("*", end=" ")
    print()

# Lower half of butterfly
for i in range(5-1, 0, -1):
    for j in range(i):  # print stars on left
        print("*", end=" ")
    for k in range(2 * (5 - i)):  # print space in middle
        print(" ", end=" ")
    for j in range(i):  # print stars on right
        print("*", end=" ")
    print()
```

```python
# increasing number triangle
for i in range(1, 6):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
```

```python
# decreasing number
for i in range(5, 0, -1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
```

```python
# increasing number with space
for i in range(1, 6):
    for j in range(5 - i):
        print(" ", end=" ")
    for k in range(1, i + 1):
        print(k, end=" ")
    print()
```

```python
# reverse counting triangle
for i in range(5, 0, -1):
    for j in range(i, 0, -1):
        print(j, end=" ")
    print()
```

```python
# increasing Alphabet triangle
for i in range(1, 6):
    for j in range(i):
        print(chr(65 + j), end=" ")
    print()
```

```python
# Decreasing Alphabet triangle
for i in range(5, 0, -1):
    for j in range(i):
        print(chr(65 + j), end=" ")
    print()
    
```

```python
# Reverse Alphabet triangle
for i in range(5, 0, -1):
    for j in range(i - 1, -1, -1):
        print(chr(65 + j), end=" ")
    print()
```

```python
# increasing Alphabet pyramid with Space
for i in range(1, 6):
    for j in range(5 - i):
        print(" ", end=" ")
    for j in range(i):
        print(chr(65 + j), end=" ")
    for j in range(i - 2, -1, -1):
        print(chr(65 + j), end=" ")
    print()
```

```python
# 1. Creating a tuple[cite: 16]
my_tuple = (1, 2, 3, 4, 5)
print("tuple:", my_tuple)
```

```python
# 2. Accessing elements in a tuple
print("First element:", my_tuple[0])   
print("last element:", my_tuple[-1]) 
```

```python
# 3. Slicing a tuple
print("Slice[1:4]:", my_tuple[1:4])                      
print("slice from begining to index 3:", my_tuple[:3])    
print("slice from index 2 to end:", my_tuple[2:])        
```

```python
# 4. Concatenating two tuples
extra_tuple = (9, 8, 7)
concatenated = my_tuple + extra_tuple
print("concatenated tuple:", concatenated)
```

```python
# 5. Iterating through a tuple
for item in my_tuple:
    print("item:", item)
```

```python
# 6. Membership test in a tuple
print("is 3 in tuple?", 3 in my_tuple)
print("is 10 not in tuple?", 10 in my_tuple)
```

```python
# 7. Nested tuple and accessing their elements
nested_tuple = (my_tuple, (7, 8, 9))
print("nested tuple:", nested_tuple)
print("Accessing nested element (my_tuple[2]):", nested_tuple[0][2])
print("Accessing nested element (9):", nested_tuple[1][2])
```

```python
# 8. Applying built-in functions on tuples
print("length of tuple:", len(my_tuple))
print("maximum value:", max(my_tuple))
print("minimum value:", min(my_tuple))
print("sum of elements:", sum(my_tuple))      
```

```python
# 9. Updating a tuple (using list conversion)
tem_list = list(my_tuple)
tem_list[2] = 9  # Update 3rd element
updated_tuple = tuple(tem_list)
print("updated tuple:", updated_tuple)
```

```python
# 10. Deleting a tuple
del my_tuple
# print(my_tuple) # Uncommenting this will cause an error because tuple is deleted
```

```python
# 1. Creating and accessing strings
my_string = "hello, edunet!"
print("original string:", my_string) 
print("first character:", my_string[0])  
print("last character:", my_string[-1])  
```

```python
# 2. String concatenation and repetition
str1 = "edunet"
str2 = "Foundation"
concat_str = str1 + " " + str2
repeated_str = concat_str * 3
print("Concatenated string:", concat_str)  
print("repeated string:", repeated_str)
```

```python
# 3. String case manipulation
upper_str = my_string.upper()         # Saare characters uppercase mein convert karna
lower_str = my_string.lower()         # Saare characters lowercase mein convert karna
title_str = my_string.title()         # Har word ka pehla letter capital karna
swapcase_str = my_string.swapcase()   # Uppercase ko lowercase aur lowercase ko uppercase karna

print("uppercase:", upper_str)       
print("lowercase:", lower_str)        
print("title case:", title_str)      
print("swap case:", swapcase_str)   
```

```python
# 4. Searching in strings
substring = "edunet"
found_index = my_string.find(substring)
if found_index != -1:
    print(f"substring '{substring}' found at index{found_index}")
else:
    print(f"substring '{substring}' not found")
```

```python
# 5. Replacing substrings
new_string = my_string.replace("edunet", "world")
print("string after replacement:", new_string)
```

```python
# 6. String formatting
name = "anjali"
age = 25
formatted_str = f"my name is {name} and I am {age} years old."
print("formatted string:", formatted_str)
```

```python
# 7. Original string with extra spaces
text = " extra spaces "

# Padding example (manually adding)
padded = "*** hello, edunet!***"
print("padded string:", padded)

# Trimming example (removing leading and trailing spaces)
trimmed = text.strip()
print("trimmed string:", trimmed)
```

```python
# 8. Splitting and joining strings
sentence = "python is fun"
words = sentence.split() # splitting based on space[cite: 15]
joined_sentence = "-".join(words) # joining with hyphen[cite: 15]
print("splitted words:", words)
print("joined sentence:", joined_sentence)
```

```python
#9. Counting characters
char_count = my_string.count("e")
print(f"Character 'e' appears {char_count} times in the string")#
```

```python
# 10. Checking string properties
is_alpha = "hello".isalpha()
is_digit = "12345".isdigit()
is_alnum = "hello123".isalnum()
is_space = " ".isspace()

print("is 'hello' alphabetic?", is_alpha)
print("is '12345' numeric?", is_digit)
print("is 'hello123' alphanumeric?", is_alnum)
print("is ' ' all space?", is_space)
```

```python
set1 = {1, 2, 3}
set2 = {3, 4, 5}
```

```python
# 1. Union of set (dono sets ke sabhi unique elements ko jodna)
union_set = set1.union(set2)
print(union_set)
```

```python
# 2. Intersection of set (dono sets ke common elements ko find karna)
intersection_set = set1.intersection(set2)
print(intersection_set)
```

```python
# 3. Difference of set (set1 ke wo elements jo set2 mein nahi hain)
difference_set = set1.difference(set2)
print(difference_set)
```

```python
# 4. Symmetric difference of sets (jo elements dono sets mein common nahi hain unhe lena)
symmetric_difference_set = set1.symmetric_difference(set2)
print(symmetric_difference_set)
```

```python
# difference of sets[cite: 14]
difference_set = set1.difference(set2)
print("difference of set A and set B(A-B):", difference_set)
```

```python
# [49]: Symmetric difference of sets[cite: 14]
sym_diff_set = set1.symmetric_difference(set2)
print("symmetric difference of set A and set B:", sym_diff_set)
```

```python
# [54]: Subset and superset testing[cite: 14]
set_c = {4, 5}
is_subset = set_c.issubset(set1)
is_superset = set1.issuperset(set_c)
print(f"is set C a subset of set A? {is_subset}")
print(f"is set A a superset of set C? {is_superset}")
```

```python
# [56]: Set membership testing[cite: 14]
is_member = 3 in set1
print("Is 3 in set A?", is_member)
```

```python
# [57]: Removing duplicates using sets[cite: 14]
numbers_list = [1, 2, 3, 4, 4, 5, 6, 6, 7]
unique_numbers = set(numbers_list)
print("list of numbers after removing duplicates:", unique_numbers)
```

```python
# [58]: Set Comprehension[cite: 15]
square_set = {x**2 for x in range(1, 11)}
print("set of square of numbers from 1 to 10:", square_set)
```

```python
# [59]: Clearing a Set[cite: 15]
set1.clear()
print("set A after clearing:", set1)
```
