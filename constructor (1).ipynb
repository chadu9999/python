```python
# non-parametrized/ default constructor
class abc:
    def __init__(self):
        print("just wow")
a = abc()
```

```python
# parametrized constructor
class abc:
    def __init__(self,a,b):
        self.a = a
        self.b = b 
        print(self.a+self.b)
a = abc(10,5)
```

```python
class abc:
    def __init__(self):
        print("hello chandni")
    def __del__(self):
        print("here i am the legend destructor")
a = abc()
del a
```

```python
class abc:
    def __init__(self):
        print("hello")
    def yes(self):
        print("world")
    def __del__(self):
        print("tata bye bye gaya")
a = abc()
del a
a.yes
```
