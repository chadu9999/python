```python
from abc import ABC, abstractmethod

# 1. Define Abstract Base Class
class BillingRecord(ABC):
    @abstractmethod
    def get_bill(self):
        """
        Abstract method to get the billing details.
        Subclasses must implement this method.
        """
        pass

    @abstractmethod
    def update_bill(self, amount):
        """
        Abstract method to update the billing amount.
        Subclasses must implement this method.
        """
        pass


# 2. Define Concrete Class
class PatientBilling(BillingRecord):
    def __init__(self, patient_id, patient_name, billing_amount=0):
        self.patient_id = patient_id
        self.patient_name = patient_name
        self.billing_amount = billing_amount

    def get_bill(self):
        return {
            'Patient ID': self.patient_id,
            'Patient Name': self.patient_name,
            'Billing Amount': f"₹{self.billing_amount:.2f}"
        }

    def update_bill(self, amount):
        if amount > 0:
            self.billing_amount += amount
            print(f"Billing amount updated by ₹{amount:.2f}. New amount is ₹{self.billing_amount:.2f}.")
        else:
            print("Amount to update must be positive.")


# 3. Use the Classes (Example Usage)
if __name__ == "__main__":
    # Create a billing record for a patient
    patient_bill = PatientBilling(
        patient_id="P001",
        patient_name="Ramar Bose",
        billing_amount=150.00
    )

    # Print initial bill details
    print("Initial Bill Details:")
    print(patient_bill.get_bill())

    # Update billing amount
    patient_bill.update_bill(50.00)  # Adding a new charge

    # Print updated bill details
    print("\nUpdated Bill Details:")
    print(patient_bill.get_bill())
```

```python
# lab 54 (a) Single Inheritance
class Employee:
    def __init__(self, name, employee_id, salary):
        """
        Initializes the Employee class with common employee details such as name, ID, and salary.
        """
        self.name = name
        self.employee_id = employee_id
        self.salary = salary

    def display_details(self):
        """
        Displays the employee's details.
        """
        return f"Employee Name: {self.name}\nEmployee ID: {self.employee_id}\nSalary: ₹{self.salary:.2f}"


class Manager(Employee):
    def __init__(self, name, employee_id, salary, team_size):
        """
        Initializes the Manager class, which inherits from Employee and adds the team size attribute.
        """
        super().__init__(name, employee_id, salary)
        self.team_size = team_size

    def display_details(self):
        """
        Displays the manager's details, including team size.
        """
        employee_details = super().display_details() # Call the base class method
        return f"{employee_details}\nTeam Size: {self.team_size}"


if __name__ == "__main__":
    # Create an instance of Employee
    employee = Employee("Employee1", "E001", 50000)

    # Create an instance of Manager
    manager = Manager("Manager", "M456", 80000, 10)

    # Display employee details
    print("Employee Details:")
    print(employee.display_details())

    print("\nManager Details:")
    # Display manager details
    print(manager.display_details())
```

```python
# multiple inheritance
class Person:
    def __init__(self, name):
        """
        Initializes the Person class with the person's name.
        """
        self.name = name

    def greet(self):
        """
        Returns a greeting message with the person's name.
        """
        return f"Hello, I'm {self.name}"


class Employee:
    def __init__(self, employee_id):
        """
        Initializes the Employee class with the employee's ID.
        """
        self.employee_id = employee_id

    def get_employee_id(self):
        """
        Returns the employee ID.
        """
        return self.employee_id


class Manager(Person, Employee):
    def __init__(self, name, employee_id):
        """
        Initializes the Manager class, which inherits from both Person and Employee
        """
        Person.__init__(self, name)
        Employee.__init__(self, employee_id)


if __name__ == "__main__":
    # Create an instance of Manager
    manager = Manager("Ramar Bose", "E12345")

    # Output manager's personal greeting
    print(manager.greet())

    # Output manager's employee ID
    print(f"Employee ID: {manager.get_employee_id()}")
```

```python
# multilevel inheritance 
class Person:
    def __init__(self, name, age):
        """
        Initializes the Person class with common personal details such as name and age.
        """
        self.name = name
        self.age = age

    def display_personal_details(self):
        """
        Displays the person's name and age.
        """
        return f"Name: {self.name}, Age: {self.age}"


class Employee(Person):
    def __init__(self, name, age, employee_id, salary):
        """
        Initializes the Employee class, which inherits from Person and adds employee details.
        """
        super().__init__(name, age) # Call the constructor of the Person class[cite: 3]
        self.employee_id = employee_id
        self.salary = salary

    def display_employee_details(self):
        """
        Displays employee-specific details such as employee ID and salary.
        """
        personal_details = self.display_personal_details() # Call the base class method[cite: 3]
        return f"{personal_details}, Employee ID: {self.employee_id}, Salary: ₹{self.salary:.2f}"


class Manager(Employee):
    def __init__(self, name, age, employee_id, salary, department, team_size):
        """
        Initializes the Manager class, which inherits from Employee and adds manager-specific details.
        """
        super().__init__(name, age, employee_id, salary) # Call the constructor of the Employee class[cite: 4]
        self.department = department
        self.team_size = team_size

    def display_manager_details(self):
        """
        Displays manager-specific details along with inherited personal and employee details.
        """
        employee_details = self.display_employee_details() # Call the derived class method[cite: 4]
        return f"{employee_details}, Department: {self.department}, Team Size: {self.team_size}"


if __name__ == "__main__":
    # Create an instance of Manager[cite: 5]
    manager = Manager("Ramar Bose", 45, "M12345", 95000, "Sales", 10)
    
    # Output manager's details[cite: 5]
    print("Manager Details:")
    print(manager.display_manager_details())
```

```python
# hierarchical inheritance 
class Person:
    def __init__(self, name, age):
        """
        Initializes the Person class with common personal details such as name and age.
        """
        self.name = name
        self.age = age

    def display_personal_details(self):
        """
        Displays the person's name and age.
        """
        return f"Name: {self.name}, Age: {self.age}"


class Employee(Person):
    def __init__(self, name, age, employee_id, salary):
        """
        Initializes the Employee class, which inherits from Person and adds employee-specific details.
        """
        super().__init__(name, age)  # Call the constructor of the Person class[cite: 7]
        self.employee_id = employee_id
        self.salary = salary

    def display_employee_details(self):
        """
        Displays employee-specific details including employee ID and salary.
        """
        personal_details = self.display_personal_details()  # Call the base class method[cite: 7]
        return f"{personal_details}, Employee ID: {self.employee_id}, Salary: ₹{self.salary:.2f}"


class Manager(Person):
    def __init__(self, name, age, department, team_size):
        """
        Initializes the Manager class, which inherits from Person and adds manager-specific details.
        """
        super().__init__(name, age)  # Call the constructor of the Person class[cite: 8]
        self.department = department
        self.team_size = team_size

    def display_manager_details(self):
        """
        Displays manager-specific details including department and team size.
        """
        personal_details = self.display_personal_details()  # Call the base class method[cite: 8]
        return f"{personal_details}, Department: {self.department}, Team Size: {self.team_size}"


if __name__ == "__main__":
    # Create an instance of Employee[cite: 9]
    employee = Employee("Employee1", 30, "E12345", 50000)
    
    # Create an instance of Manager[cite: 9]
    manager = Manager("JManager", 40, "Sales", 12)
    
    # Output employee's details[cite: 9]
    print("Employee Details:")
    print(employee.display_employee_details())
    
    # Output manager's details[cite: 9]
    print("\nManager Details:")
    print(manager.display_manager_details())
```

```python
#  method overloading
class printer:
    def print_message(self, message,times=1):
        for _ in range(times):
            print(message)
if __name__ == "__main__":
    printer = printer()
    printer.print_message("hello")   # default to 1 time 
    printer.print_message("hello",3)   # print 3 times
```

```python
# method overriding
class animal:
    def speak(self):
        return "animal speaks"
class dog(animal):
    def speak(self):
        return "woof"
class cat(animal):
    def speak(self):
        return "meow"
def make_animal_speak(animal):
    print(animal.speak())
if __name__ == "__main__":
    dog = dog()
    cat = cat()

    make_animal_speak(dog)
    make_animal_speak(cat)
```
