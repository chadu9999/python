```python
def divide_else (num1, num2):
    try:
        print(num1 / num2)
    except ZeroDivisionError as e:
        print (' catch ZeroDivisionError:' , e)
    else:
        print('finish(no error)')

divide_else(1,2)
```

```python
def divide_finally (a, b):
    try:
        print(a / b)
    except ZeroDivisionError as e:
        print (' catch ZeroDivisionError:' , e)
    finally:
        print(' finish(no error)')

divide_finally(1,0)
```
