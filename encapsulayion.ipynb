```python
class student:
    def __init__(self, name, marks):
        self.name = name     # public attibute
        self.marks = marks    # public attibute
    def display(self):        # public method 
        print("Name:", self.name)
        print("Marks:", self.marks)
# creating an object
s1 = student("chandni", 90)
#accessing public attributes
print(s1.name)
print(s1.marks)

# calling a public method
s1.display()
```

```python
class director:
    def __init__(self, name, salary):
        self.name = name         # public attibute
        self._salary = salary   # protected attibute
        
    def _display_salary(self):          # protected method 
        print("salary:", self._salary)


class manager(director):
    def show_details(self):
        print("name:", self.name)

        #accessing the protected attribute
        print("salary:", self._salary)
        
       #calling the protected method
        self._display_salary()
#creating an object of the child class 
m1 = manager("chandni", 60000)

m1.show_details()
```

```python
class bankaccount:
    def __init__(self, name, balance, pin):
        self.name = name 
        self.__balance = balance
        self.__pin = pin
    def __verify_pin(self, entered_pin):
        return self.__pin == entered_pin 

    def show_balance(self, entered_pin):
        if self.__verify_pin(entered_pin):
            print("balance:", self.__balance)
        else:
            print("incorrect Pin")
# creating an object 
account = bankaccount("chandni", 10000, 2007)
# accessing A PUBLIC ATTRIBUTE
print(account.name)


account.show_balance(2007)


print(account.__balance)
```

```python
class person:
    def __init__(self,name):
        self.__name = name
    def get_name(self):
        return self.__name

    def set_name(self, name):
        self.__name = name

person = person("shalu")

print(person.get_name())

person.set_name("chandni")

print(person.get_name())
```

```python
from abc import ABC, abstractmethod 
# abstract class 
class student(ABC):

#abstract method to calculate grade 
   @abstractmethod
   def calculate_grade(self):
       pass

# deriived class for undergradute students

class UndergraduateStudent(student):
    def __init__(self,marks):
        self.marks = marks
    def calculate_grade(self):
        if self.marks >= 60:
            return "pass"
        else:
            return "fail"

# derived class for graduate students
class Graduatestudent(student):
    def __init__(self,marks):
        self.marks = marks 

    def calculate_grade(self):
        if self.marks >= 70:
           return"pass"
        else:
            return"fail"

#creating object of derived classes
undergrad = UndergraduateStudent(60)
grad = Graduatestudent(68)
# accessing the abstracted method to calculate grades
print("undergraduate student grade:", undergrad.calculate_grade())
print ("graduate student grade:", grad.calculate_grade())
        
```
