```python
# Polymorphism in Python
# operator  overloading 

class vector:
    def __init__(self, num1, num2):
        # assign the
        self.x = num1 
        self.y = num2
    def __add__(self, other):
        return vector(self.x + other.x, self.y + other.y)
    def __str__(self):
        return f"vector({self.x}, {self.y})"
v1 = vector(2,3)
v2 = vector(4,5) 

result = v1 + v2
print(result)
```
