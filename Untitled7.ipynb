```python
# Writing binary data to a file
file = open('File.dat', 'wb') # Open in write binary mode
file.write(b"Hello, Edunet") # Write the binary data as bytes
file.close() # Close the file

# Reading binary data from the file
file = open('File.dat', 'rb') # Open in read binary mode
content = file.read() # Read the entire binary content
print(content) # Output the binary data
file.close() # Close the file
```
