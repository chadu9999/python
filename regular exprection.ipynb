```python
import re
pattern = r'\d+'
string = '123 abc'

match = re.match(pattern,string)
if match:
    print("Match found:", match.group())
```

```python
import re
pattern = r'\d*'
string = 'abc'

match = re.match(pattern,string)
if match:
    print("Match found:", match.group())
```

```python
#Example 1: Find a particular word
import re

text = "Python is easy. Python is powerful."

result = re.findall(r"Python", text)

print(result)
```

```python
#Example 2: Find a word without case sensitivity
import re

text = "Python python PYTHON"

result = re.findall(r"python", text, re.IGNORECASE)

print(result)
```

```python
#Example 3: Find all occurrences of a word
import re

text = "AI is growing. AI is useful. AI is powerful."

result = re.findall(r"AI", text)

print(result)
```

```python
#Example 4: Find words that start with A.
import re

text = "Apple Ant Ball Aeroplane Cat"

result = re.findall(r"\bA\w*", text)

print(result)
```

```python
#3. Metacharacters
# Example 1: Match any character in a position.
import re

text = "cat cot cut"

result = re.findall(r"c.t", text)

print(result)
```

```python
# Example 2: Match any three characters.
import re

text = "cat dog sun"

result = re.findall(r"...", text)

print(result)
```

```python
# Example 3: Match a dot literally
import re

text = "file.txt file.csv file.pdf"

result = re.findall(r"\.txt", text)

print(result)
```

```python
# 3.2 Caret (^): Beginning of a String
# Example 1: Check whether a string starts with Python
import re

text = "Python is easy"

result = re.search(r"^Python", text)

print(result.group() if result else "No match")
```

```python
# Example 2: Check a string that does not start with Python
import re

text = "I am learning Python"

result = re.search(r"^Python", text)

print(result.group() if result else "No match")
```

```python
# Example 3: Find words beginning with A
import re

text = "Apple Ant Ball Aeroplane"

result = re.findall(r"\bA\w*", text)

print(result)
```

```python
# 3.3 Dollar ($): End of a String
# Example 1: Check whether the string ends with Python
import re

text = "I am learning Python"

result = re.search(r"Python$", text)

print(result.group() if result else "No match")
```

```python
# Example 2: Check whether the string ends with Java
import re

text = "I am learning Python"

result = re.search(r"Java$", text)

print(result.group() if result else "No match")
```

```python
# Example 3: Find a three-digit string
import re

print(bool(re.fullmatch(r"\d\d\d", "123")))
print(bool(re.fullmatch(r"\d\d\d", "1234")))
```

```python
# 3.4 Vertical Bar (|): OR Operator
# Example 1: Match cat or dog
import re

text = "cat dog cow"

result = re.findall(r"cat|dog", text)

print(result)
```

```python
# Example 2: Match multiple alternatives 
import re

text = "I use Python, Java, SQL and C++."

result = re.findall(r"Python|Java|SQL", text)

print(result)
```

```python
# Example 3: Match Mr or Mrs 
import re

text = "Mr Sharma and Mrs Sharma"

result = re.findall(r"Mr|Mrs", text)

print(result)
```

```python
# Notice that the Mr alternative also matches the beginning of Mrs in a larger expression like this. A more appropriate pattern for matching either title is:
import re

text = "Mr Sharma and Mrs Sharma"

result = re.findall(r"Mrs|Mr", text)

print(result)
```

```python
# 3.5 Parentheses (()): Grouping
# Example 1: Group alternatives
import re

text = "I like apples and oranges"

result = re.findall(r"(apples|oranges)", text)

print(result)
```

```python
# Example 2: Capture a year
import re

text = "The event happened in 2026."

result = re.search(r"(\d{4})", text)

print(result.group(1))
```

```python
# Example 3: Use a non-capturing group
import re

text = "cat dog"

result = re.findall(r"(?:cat|dog)", text)

print(result)
```

```python
# 3.6 Backslash (\): Escape Character
# Example 1: Match a digit
import re

text = "My age is 27"

result = re.findall(r"\d", text)

print(result)
```

```python
# Example 2: Match an actual plus sign
import re

text = "Python C++ Java"

result = re.findall(r"C\+\+", text)

print(result)
```

```python
# Example 3: Match an actual dollar sign
import re

text = "Price is $500"

result = re.findall(r"\$\d+", text)

print(result)
```

```python
# 4. Character Classes
# Square Brackets ([])
# Example 1: Match a, b, or c
import re

text = "apple ball cat dog"

result = re.findall(r"[abc]", text)

print(result)
```

```python
# Example 2: Match vowels 
import re

text = "education"

result = re.findall(r"[aeiou]", text)

print(result)
```

```python
# Example 3: Match uppercase letters
import re

text = "Python AI SQL"

result = re.findall(r"[A-Z]", text)

print(result)
```

```python
#Example 2: Match one or more lowercase letters
import re

text = "abc123xyz"

result = re.findall(r"[a-z]+", text)

print(result)
```

```python
# Example 3: Match repeated a characters
import re

text = "a aa aaa b"

result = re.findall(r"a+", text)

print(result)
```

```python
#Question Mark (?): Zero or One
import re

text = "color colour"

result = re.findall(r"colou?r", text)

print(result)
```
