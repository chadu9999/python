```python
# instance method

class student:
    def __init__(self, name, age):
        self.name = name   # instance variable
        self.age = age   # instance variable
# creating objects
s1 = student("rahul", 20)
s2 = student("priya", 22)
print(s1.name)
print(s1.age)

print(s2.name)
print(s2.age)
```

```python
# class method 

class student:
    school_name = "someswerpuri"  # class variable

    def __init__(self, name):
        self.name = name   # instance variable


s1 = student("rahul")
s2 = student("priya")
print(s1.name)
print(s1.school_name)

print(s2.name)
print(s2.school_name)

```

```python
# static method

class student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    def display(self):    # instance method 
        print("name:", self.name)
        print("marks:", self.marks)

s1 = student("rahul", 85)
s2 = student("priya", 92)


s1.display()
s2.display()

```

```python
# class method 
class student:
    school_name = "bhagwati school"

    def __init__(self, name):
        self.name = name 
    @classmethod
    def change_school(cls,new_name):
        cls.school_name = new_name
s1 = student("rahul")
s2 = student("priya")

print(s1.school_name)
print(s2.school_name)

student.change_school("someswerpuri school")

print(s1.school_name)
print(s2.school_name)
```

```python
# static method

class student:

    @staticmethod
    def is_passed(marks):
        if marks >=40:
            return True
        else:
            return False

print(student.is_passed(75))
print(student.is_passed(30))

```
