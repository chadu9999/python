```python
# single inheritance
class parent:
    def add (self,a,b):
        print(a+b)

class child(parent):
    def add(self):
        print("wow")
        c=super().add(5,10)
c = child()
c.add()
p = parent()
p.add(5,10)
```

```python
class parent1:     #multiple inheritance 
    def add (self, a,b):
        print (a+b)

class parent2:
    def sub(self,c,d):
        print(c-d)
class child (parent1,parent2):
    def mul(self,e,f):
        print(e*f)

c = child()
c.add(5,10)
c.sub(10,5)
c.mul(5,50)

```

```python
class parent:    # hierarchical inheritance
    def add (self, a,b):
        print (a+b)

class child1(parent):
    def sub(self,c,d):
        print(c-d)
class child2 (parent):
    def mul(self,e,f):
        print(e*f)

c1 = child1()
c1.add(5,10)
c1.sub(10,5)

c2 = child2()
c2.add(50,10)
c2.mul(50,50000)
```
